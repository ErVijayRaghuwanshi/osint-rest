```python
from requests import Session
from requests import RequestException
import json

header_path = "../config/headers.json"

with open(header_path, "r") as f:
    headers = json.load(f)['platforms']['jaco']['headers']

s = Session()
s.headers = headers[0]['headers']


def get_user_info(username:str='Uaegirl'):
    resp = s.get(f"https://api.jaco.live/search/c/offical/associate.json?keyword={username}&show_agent=0").json()

    id = resp['data']['list'][0]['content']['user']['id']
    if username == resp['data']['list'][0]['content']['user']['username']:
        url = f"https://api.jaco.live/bff/profile/header/user/web?uid={id}&show_badge=1&show_count=1&show_tab=1&show_business=1&show_live=1"

        response = s.get(url)
        return response.json()
    else:
        raise "User not found"
user_info = get_user_info("Uaegirl")
user_info
```

```python


def user_feed(uid=3004008323, limit:int=12, next_since_id:int = None):
    feed = {}
    try:
        while True:
            params = {'target_uid':uid, 'next_since_id': next_since_id}
            resp = s.get("https://api.jaco.live/reel/2/reel/c/flow.json", params=params)
            if resp.ok:
                data = resp.json()
                if feed:
                    feed['data']['list'].extend(data['data']['list'])
                else:
                    feed = data
                params['next_since_id'] = data['data']['next_since_id']
                feed['data']['next_since_id'] = data['data']['next_since_id']
                feed['data']['pre_since_id'] = data['data']['pre_since_id']
                print(len(feed['data']['list']))
                if len(feed['data']['list']) >= limit:
                    break
            else:
                resp.raise_for_status()
                break
    except Exception as e:
        print(e)
    finally:
        return feed

data = user_feed(uid=3004008323, limit=20)
            
```
